```python
def create_acc(accounts):
    account_number = int(input("Enter account number: ")) 
    
    # Check if account exists
    if any(acc["account_number"] == account_number for acc in accounts):
        print("Account number already exists. Try creating with a new one.")
        return  

    customer_name = input("Enter customer name: ")
    balance = int(input("Enter initial balance: "))
    account_type = input("Enter account type (Savings/Current): ")

    account = {
        "account_number": account_number,
        "customer_name": customer_name,
        "balance": balance,
        "account_type": account_type
    } 
    
    accounts.append(account)
    print(accounts)
    print("Account created successfully!")


def deposit_money(accounts):
    print("DEPOSIT")
    account_number = int(input("Enter account number: "))
    for acc in accounts:
        
        amount = int(input("Enter the amount to be deposited: "))
        if amount<=0:
                print("Deposit amount must be greater than 0")
                return
        elif acc["account_number"] == account_number:
            acc["balance"] += amount
            print(f"Deposited successfully. New balance: {acc['balance']}")
            return

        '''elif acc["account_number"] not in accounts :
                print("Enter EXISTING/CORRECT ACCOUNT NUMBER")
                return'''
        
    else:
                  print("ACCOUNT DOES NOT EXIST")
                  return  
    
def withdraw_money(accounts):
        account_number = int(input("Enter account number: "))
        for acc in accounts:
            amount = int(input("Enter the amount to be withdrawn:"))
            if acc["account_number"] == account_number:
                if amount<= acc["balance"]:   
                    acc["balance"] -= amount
                    print(f"Withdrawn successfully. New balance: {acc['balance']}")
                    #print(accounts)
                    return
            elif acc["account_number"] not in accounts:
                print("Enter EXISTING ACCOUNT NUMBER")
                return
        else:
              print("ACCOUNT DOES NOT EXIST")
              return    
                
def check_balance(accounts) :
     account_number = int(input("Enter the account number that you want to check the balance for"))
     for acc in accounts:
        print("CHECKING BALANCE")
        if acc["account_number"] == account_number:
            print("The balance in account",acc["customer_name"],"'s account is",acc["balance"])
            return
        

def transfer_money(accounts):
    sender_accnum =  int(input("Enter the sender's account number"))
    receiver_accnum = int(input("Enter Receiver account number: "))
    
    sender_acc = None
    for acc in accounts:
         if acc['account_number'] == sender_accnum:
              sender_acc = acc
              print('SENDER ACC:',sender_acc)
                  
    receiver_acc = None
    for acc in accounts:
             if acc['account_number'] == receiver_accnum:
                  receiver_acc = acc
                  print('RECIEVER ACC:',receiver_acc)     
    
    if sender_acc is None or receiver_acc is None:
         print("Account Not found")
         
    if sender_accnum == receiver_accnum:
         print("The sender and reciever cannot be the same person")
         return 
         
    #print("The sender's number is",sender_acc,"and the reciever's number is",receiver_acc)
    
    amount = int(input("Enter transferable amount"))
    if amount>=sender_acc['balance']:
         print("The amount to be transferred cannot exceed available balance from the sender's account")
         return

    
    sender_acc['balance']-=amount
    receiver_acc['balance']+=amount
    print("The recieved amount from",sender_acc['customer_name'],"'s account is"+str(amount))
    #print("The balance of"+receiver_accnum['customer_name']+"'s account after recieveing Rs",amount,"is",receiver_accnum["balance"])




         
              






def main():
    # This list is created ONCE at startup
    accounts = [] 
    print("\n--- Banking Menu ---")
    print("1. Create Account")
    print("2. Deposit Money")
    print("3.Withdraw Money ")
    print("4.Check balance")
    print("5.Transfer Money")
    print("9. Exit")
            
    while True: 
        
        # CRITICAL: This input must be INSIDE the loop so it asks you again 
        # WITHOUT restarting the whole script or wiping the list.
        user_op = input("What operation do you want to do? ")   
        
        if user_op == '1':  
            create_acc(accounts)
        elif user_op == '2':
            deposit_money(accounts)
        elif user_op == '3':
            withdraw_money(accounts)    
        elif user_op == '4':
            check_balance(accounts) 
        elif user_op == '5':
             transfer_money(accounts)   
        elif user_op == '9':
            print("Exiting application...")
            break # This stops the program safely
        else:
            print("Invalid option, please try again.")
            
    print("\nThe final Accounts enlisted are:", accounts)

# Run the program exactly once
main()
```
